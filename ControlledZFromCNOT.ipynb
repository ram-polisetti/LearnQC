```python
from qiskit import *
from qiskit.circuit import Gate
```

```python
cz_gate = QuantumCircuit(2)
cz_gate.cz(0, 1)
```

```python
cz_gate.draw('mpl')
```

```python
# HXH = Z
# HZH = X
```

```python
z_gate = QuantumCircuit(2)
c, t = 0, 1
```

```python
z_gate.h(t)
z_gate.cx(c,t)
z_gate.h(t)
```

```python
z_gate.draw('mpl')
```

when q0 is 0 nothing happens -> HIH is H**2  and q1 remains unchanged

but when q0 is 1 HXH acts as a Z gate 

and this is the property of a cz gate
